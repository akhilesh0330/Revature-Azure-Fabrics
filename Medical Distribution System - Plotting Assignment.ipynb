```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Creating DataFrame

medical_data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Medicine_Distributed': [1200, 1500, 1800, 1700, 2100, 2500],
    'Hospital_Orders': [30, 45, 50, 48, 60, 75],
    'Delivery_Time': [5, 4, 6, 5, 3, 2],
    'Revenue': [50000, 65000, 72000, 70000, 85000, 95000]
}

df = pd.DataFrame(medical_data)

print(df)
```

Line Chart

```python
# 1. Line chart showing medicine distribution over months

plt.figure(figsize=(8,5))
plt.plot(df['Month'], df['Medicine_Distributed'],
         marker='o', color='blue')
plt.title('Medicine Distribution Over Months')
plt.xlabel('Month')
plt.ylabel('Medicine Distributed')
plt.grid(True)
plt.savefig('line_medicine_distribution.png')
plt.show()
```

```python
# 2. Line chart for monthly revenue growth

plt.figure(figsize=(8,5))
plt.plot(df['Month'], df['Revenue'],
         marker='s', color='green')
plt.title('Monthly Revenue Growth')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.grid(True)
plt.savefig('line_revenue_growth.png')
plt.show()
```

```python
# 3. Multiple line charts for Medicine_Distributed and Hospital_Orders

plt.figure(figsize=(8,5))
plt.plot(df['Month'], df['Medicine_Distributed'],
         marker='o', label='Medicine Distributed')
plt.plot(df['Month'], df['Hospital_Orders'],
         marker='s', label='Hospital Orders')
plt.title('Medicine Distribution vs Hospital Orders')
plt.xlabel('Month')
plt.ylabel('Count')
plt.legend()
plt.grid(True)
plt.savefig('multiple_line_chart.png')
plt.show()
```

BAR CHART PROBLEMS

```python
# 4. Vertical bar chart for hospital orders by month

plt.figure(figsize=(8,5))
plt.bar(df['Month'], df['Hospital_Orders'],
        color='orange')
plt.title('Hospital Orders by Month')
plt.xlabel('Month')
plt.ylabel('Orders')
plt.savefig('vertical_bar_orders.png')
plt.show()
```

```python
# 5. Horizontal bar chart for revenue by month

plt.figure(figsize=(8,5))
plt.barh(df['Month'], df['Revenue'],
         color='purple')
plt.title('Revenue by Month')
plt.xlabel('Revenue')
plt.ylabel('Month')
plt.savefig('horizontal_bar_revenue.png')
plt.show()
```

```python
# 6. Grouped bar chart comparing Medicine_Distributed and Hospital_Orders

x = np.arange(len(df['Month']))
width = 0.35

plt.figure(figsize=(9,5))

plt.bar(x - width/2,
        df['Medicine_Distributed'],
        width,
        label='Medicine Distributed')

plt.bar(x + width/2,
        df['Hospital_Orders'],
        width,
        label='Hospital Orders')

plt.xticks(x, df['Month'])
plt.title('Medicine Distributed vs Hospital Orders')
plt.xlabel('Month')
plt.ylabel('Values')
plt.legend()

plt.savefig('grouped_bar_chart.png')
plt.show()
```

SCATTER PLOT PROBLEMS

```python
# 7. Scatter plot: Hospital Orders vs Revenue

plt.figure(figsize=(8,5))
plt.scatter(df['Hospital_Orders'],
            df['Revenue'],
            color='red',
            s=100)

plt.title('Hospital Orders vs Revenue')
plt.xlabel('Hospital Orders')
plt.ylabel('Revenue')
plt.grid(True)

plt.savefig('scatter_orders_revenue.png')
plt.show()
```

```python
# 8. Scatter plot: Delivery Time vs Revenue

plt.figure(figsize=(8,5))
plt.scatter(df['Delivery_Time'],
            df['Revenue'],
            color='green',
            s=100)

plt.title('Delivery Time vs Revenue')
plt.xlabel('Delivery Time')
plt.ylabel('Revenue')
plt.grid(True)

plt.savefig('scatter_delivery_revenue.png')
plt.show()
```

```python
# 9. Scatter plot: Medicine Distributed vs Hospital Orders

plt.figure(figsize=(8,5))
plt.scatter(df['Medicine_Distributed'],
            df['Hospital_Orders'],
            color='blue',
            s=100)

plt.title('Medicine Distributed vs Hospital Orders')
plt.xlabel('Medicine Distributed')
plt.ylabel('Hospital Orders')
plt.grid(True)

plt.savefig('scatter_medicine_orders.png')
plt.show()
```

Relationship Analysis

```python
print("Relationship Analysis:")
print("1. Hospital Orders vs Revenue -> Positive Relationship")
print("2. Delivery Time vs Revenue -> Negative Relationship")
print("3. Medicine Distributed vs Hospital Orders -> Positive Relationship")
```

PIE CHART PROBLEMS

```python
# 10. Pie chart showing medicine distribution percentage by month

plt.figure(figsize=(7,7))

plt.pie(df['Medicine_Distributed'],
        labels=df['Month'],
        autopct='%1.1f%%',
        startangle=90)

plt.title('Medicine Distribution Percentage by Month')

plt.savefig('pie_medicine_distribution.png')
plt.show()
```

```python
# 11. Pie chart for revenue contribution by month
# Highlight highest revenue month using explode

highest_index = df['Revenue'].idxmax()

explode = [0]*len(df)
explode[highest_index] = 0.2

plt.figure(figsize=(7,7))

plt.pie(df['Revenue'],
        labels=df['Month'],
        autopct='%1.1f%%',
        explode=explode,
        startangle=90)

plt.title('Revenue Contribution by Month')

plt.savefig('pie_revenue_contribution.png')
plt.show()
```
