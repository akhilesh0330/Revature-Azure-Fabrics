```python
# List of numbers
numbers = [45, 12, 78, 34, 78, 90, 12, 56, 23, 45]
```

**1. Array Sum**
Write a function that accepts a list of numbers and returns the sum of all elements.

```python
def array_sum(arr):
    total = 0

    for num in arr:
        total += num

    return total

print("Sum of Array:", array_sum(numbers))
```

**2. Maximum Element**
Write a function to find the largest element in a list without using max().

```python
def find_max(arr):
    largest = arr[0]

    for num in arr:
        if num > largest:
            largest = num

    return largest

print("Maximum Element:", find_max(numbers))
```

**3. Even Numbers Counter**
Write a function that counts how many even numbers are present in a list.

```python
def count_even(arr):
    count = 0

    for num in arr:
        if num % 2 == 0:
            count += 1

    return count

print("Even Numbers Count:", count_even(numbers))
```

**4. Reverse Array**
Write a function to reverse the elements of a list.

```python
def reverse_array(arr):
    reversed_list = arr[::-1]
    return reversed_list

print("Reversed Array:", reverse_array(numbers))
```

**5. Search Element**
Write a function to search for a given element in a list and return its index.

```python
def search_element(arr, target):

    for index in range(len(arr)):
        if arr[index] == target:
            return index

    return -1

element = 90

result = search_element(numbers, element)

if result != -1:
    print(f"Element {element} found at index {result}")
else:
    print(f"Element {element} not found")
```

**6. Palindrome List**
Write a function to check whether a list reads the same forward and backward.

```python
def is_palindrome(arr):

    if arr == arr[::-1]:
        return True
    else:
        return False

print("Is Palindrome:", is_palindrome(numbers))
```

**7. Second Largest Number**
Write a function to find the second largest number in a list.

```python
def second_largest(arr):

    largest = second = arr[0]

    for num in arr:

        if num > largest:
            second = largest
            largest = num

        elif num > second and num != largest:
            second = num

    return second

print("Second Largest Number:", second_largest(numbers))
```

**8. Duplicate Elements**
Write a function that identifies duplicate elements in a list.

```python
def find_duplicates(arr):

    duplicates = []

    for num in arr:

        if arr.count(num) > 1 and num not in duplicates:
            duplicates.append(num)

    return duplicates

print("Duplicate Elements:", find_duplicates(numbers))
```

**9. Function for Student Marks**
Create a function that accepts student marks in a list and returns average, highest, and lowest marks.

```python
student_marks = [78, 85, 92, 67, 88]

def marks_analysis(marks):

    total = sum(marks)
    average = total / len(marks)

    highest = max(marks)
    lowest = min(marks)

    return average, highest, lowest

avg, high, low = marks_analysis(student_marks)

print("Student Marks Analysis")
print("Average Marks:", avg)
print("Highest Marks:", high)
print("Lowest Marks:", low)
```
